```python
input_hours = input("Enter Hours: ")
input_rate = input("Enter Rate: ")
hours = float(input_hours)
rate = float(input_rate)  

if hours > 40:
    straight_pay = 40 * rate
    overtime_pay = (hours - 40) * (rate * 1.5)
    pay = straight_pay + overtime_pay
else:
    pay = rate * hours

print("Pay:", pay)
```

```python
inp = input("Enter score: ")

try:
    score = float(inp)

    if score >= 0.9 and score <= 1.0:
        print("A")
    elif score >= 0.8 and score < 0.9:
        print ("B")
    elif score >= 0.7 and score < 0.8:
        print("C")
    elif score >= 0.6 and score < 0.7:
        print("D")
    elif score >= 0.0 and score < 0.6:
        print("F")
    else:
        print("Bad Score")
except:
    print("Bad Score")
```

```python
inp1 = input("Enter Bond Price: ")
inp2 = input("Enter Face Value")
try:
    price = float(inp1)
    face_value = float(inp2)
    perc_of_par = (price / face_value) * 100
    
    if perc_of_par > 100:
        print("Premium")
    elif perc_of_par == 100:
        print("Par")
    else: 
        print ("Discount")
except:
    print("Incorrect entry, Please enter bond price and face value as dollar amounts. ")
```
