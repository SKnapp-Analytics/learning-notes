```python
# Exercise 6:

def compute_pay(hours, rate):

    if hours > 40:
        straight_pay = 40 * rate
        overtime_pay = (hours - 40) * (rate * 1.5)
        return(straight_pay + overtime_pay)
    else:
        return(hours * rate)


print(compute_pay(45,10))

```

```python
# Exercise 7

def computegrade(score):

    try: 
        score = float(score)

        if score >= 0.9 and score <= 1.0:
            return("A")
        elif score >= 0.8 and score < 0.9:
            return("B")
        elif score >= 0.7 and score < 0.8:
            return("C")
        elif score >= 0.6 and score < 0.7:
            return("D")
        elif score >= 0.0 and score < 0.6:
            return("F")
        else:
            return("Bad Score")
    
    except:
        return("Bad Score")

print(computegrade(10))

```

```python
# Exercise 1 & 2:

total = 0
count = 0
max = None
min = None

while True: 
    num = input('Enter a number: ')
    if num == 'done':
        break
    try:
        num_inp = float(num)
        count += 1
        total = total + num_inp               ## interates the sum
        if max is None or num_inp > max:
            max = num_inp
        if min is None or num_inp < min:
            min = num_inp
    except ValueError:
        print('Invalid Input')
if count > 0:
    print(total, count, max, min)
else:
    print('No data entered')
```
